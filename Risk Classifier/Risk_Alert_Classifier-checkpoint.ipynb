# 🚨 Risk Alert Classifier


## Part A: Conceptual Understanding (Theory)











**1. What is Logistic Regression and why is it suitable for classification?**
Logistic Regression takes a weighted sum of the features and passes it through the *sigmoid* function, which gives a probability between 0 and 1 that the customer is high-risk. If probability ≥ 0.5 we predict class 1. It is suitable because it outputs probabilities, is fast, and its coefficients are easy to explain to business teams.

**2. What are classification performance metrics and why is accuracy alone insufficient?**
They measure how well predicted classes match real classes (confusion matrix, precision, recall, F1, ROC-AUC). Accuracy is misleading on imbalanced data: if only 12% of customers are high-risk, a model that says "everyone is low-risk" gets ~88% accuracy but catches **zero** risky customers.


**3. Type-I and Type-II error in risk prediction**
- **Type-I Error (False Positive):** a low-risk customer is flagged as high-risk → unnecessary checks, an unhappy good customer.
- **Type-II Error (False Negative):** a high-risk customer is predicted low-risk → missed default/fraud, direct financial loss. This is usually the **costlier** error for a bank.

**4. Precision, Recall, F1-Score, TPR and FPR**
- **Precision** = TP / (TP + FP): of all customers we flagged, how many were truly high-risk.
- **Recall = TPR** = TP / (TP + FN): of all truly high-risk customers, how many we caught.
- **F1-Score** = 2 × (Precision × Recall) / (Precision + Recall): balance of precision and recall.
- **FPR** = FP / (FP + TN): share of low-risk customers wrongly flagged.

**5. What is AUC-ROC and how does it help?**
The ROC curve plots TPR against FPR for every possible threshold. AUC is the area under that curve: 0.5 = random guessing, 1.0 = perfect. It measures how well the model *ranks* risky customers above safe ones, independent of any single threshold, so it is great for comparing models.

**6. Why does imbalanced data create problems?**
The model sees very few minority examples, so to minimise overall error it simply favours the majority class. The result is high accuracy but very low recall for the high-risk class, i.e. the very customers we care about are missed.

## Setup: Import Libraries

```python
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, RandomizedSearchCV, GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (confusion_matrix, accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, roc_curve, classification_report)

from imblearn.over_sampling import RandomOverSampler, SMOTE, ADASYN
from imblearn.under_sampling import RandomUnderSampler

```

## Part B: Dataset Understanding & Preparation

### Task 7: Identify input features and target variable

```python
df = pd.read_csv("Risk_Alert_Classifier_Data.csv")
print("Shape:", df.shape)
display(df.head())
```

```python
# Target variable
target = "risk_status"        

# Create a useful feature from the date column: days since the customer's last transaction
df["last_transaction_date"] = pd.to_datetime(df["last_transaction_date"])
df["days_since_last_txn"] = (df["last_transaction_date"].max() - df["last_transaction_date"]).dt.days

# customer_id is just an ID and the raw date is already converted -> drop them
X = df.drop(columns=[target, "customer_id", "last_transaction_date"])
y = df[target]

print("Target variable :", target)
print("Input features  :", X.shape[1])
print(list(X.columns))
```

**Class distribution of the target** (this shows the imbalance)

```python
counts = y.value_counts()
print(counts)
print("\nHigh-risk share: {:.1f}%".format(100 * y.mean()))

plt.figure(figsize=(5, 3.5))
sns.countplot(x=y)
plt.xticks([0, 1], ["Low Risk (0)", "High Risk (1)"])
plt.title("Class Distribution of Risk Status")

plt.show()
```

**Interpretation:** only a small minority of customers are high-risk, so the dataset is clearly imbalanced.

### Task 8: Train-test split (stratified, keeps class distribution)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42)

print("Train:", X_train.shape, "| Test:", X_test.shape)
print("High-risk % in train: {:.2f}".format(100 * y_train.mean()))
print("High-risk % in test : {:.2f}".format(100 * y_test.mean()))
```

`stratify=y` makes sure train and test have the same share of high-risk customers.

### Task 9: Missing values and KNN Imputer

```python
missing = X_train.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print("Missing values in training data:")
display(missing.to_frame("missing_count").assign(percent=(missing / len(X_train) * 100).round(2)))
```

**Plan**
- **Numeric columns:** scale (StandardScaler) → fill with **KNN Imputer** (uses the 5 most similar customers; scaling is needed because KNN is distance-based).
- **Categorical columns** (`region`, `employment_type`): fill with the most frequent value, then one-hot encode.
- Everything is **fitted on the training data only** to avoid data leakage.

```python
num_cols = X_train.select_dtypes(include="number").columns.tolist()
cat_cols = [c for c in X_train.columns if c not in num_cols]

num_pipe = Pipeline([("scale", StandardScaler()),
                     ("knn_imputer", KNNImputer(n_neighbors=5))])
cat_pipe = Pipeline([("fill", SimpleImputer(strategy="most_frequent")),
                     ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))])

preprocess = ColumnTransformer([("num", num_pipe, num_cols),
                                ("cat", cat_pipe, cat_cols)])

X_train_p = preprocess.fit_transform(X_train)   
X_test_p  = preprocess.transform(X_test)        

feature_names = preprocess.get_feature_names_out()
print("Processed train shape:", X_train_p.shape)
print("Missing values left (train, test):", np.isnan(X_train_p).sum(), np.isnan(X_test_p).sum())
```

### Helper function
One small function to evaluate every model the same way and store the results for the final comparison.

```python
results = {}     
roc_store = {}   

def evaluate(name, model, X_te=None, y_te=None):
    """Evaluate a fitted model on the test set and store the metrics."""
    X_te = X_test_p if X_te is None else X_te
    y_te = y_test   if y_te is None else y_te
    pred  = model.predict(X_te)
    proba = model.predict_proba(X_te)[:, 1]
    tn, fp, fn, tp = confusion_matrix(y_te, pred).ravel()
    auc = roc_auc_score(y_te, proba)
    results[name] = {
        "Accuracy": accuracy_score(y_te, pred),
        "Precision": precision_score(y_te, pred, zero_division=0),
        "Recall (High Risk)": recall_score(y_te, pred),
        "F1-Score": f1_score(y_te, pred),
        "AUC-ROC": auc,
        "TN": tn, "FP (Type-I)": fp, "FN (Type-II)": fn, "TP": tp}
    fpr, tpr, _ = roc_curve(y_te, proba)
    roc_store[name] = (fpr, tpr, auc)
    return results[name]

def plot_confusion(model, title, fname=None):
    pred = model.predict(X_test_p)
    cm = confusion_matrix(y_test, pred)
    plt.figure(figsize=(4.2, 3.6))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False,
                xticklabels=["Pred Low", "Pred High"], yticklabels=["Actual Low", "Actual High"])
    plt.title(title)
    plt.show()
```

## Part C: Baseline Classification Model

### Task 10: Logistic Regression baseline

```python
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train_p, y_train)
evaluate("Logistic Regression (baseline)", log_reg)
print("Baseline model trained")
```

### Task 11: Confusion matrix, accuracy, precision, recall, F1

```python
plot_confusion(log_reg, "Baseline Logistic Regression", "02_baseline_confusion_matrix")
print(classification_report(y_test, log_reg.predict(X_test_p), target_names=["Low Risk", "High Risk"]))
display(pd.DataFrame(results).T[["Accuracy", "Precision", "Recall (High Risk)", "F1-Score", "AUC-ROC"]].round(3))
```

**Interpretation:** the baseline Logistic Regression is already extremely strong on this dataset (near-perfect on every metric). This is unusual for real banking data, so before trusting it we run a sanity check below.

```python
from sklearn.model_selection import cross_val_score
cv_auc = cross_val_score(LogisticRegression(max_iter=1000), X_train_p, y_train,
                         cv=StratifiedKFold(5, shuffle=True, random_state=42), scoring="roc_auc")
print("5-fold CV AUC-ROC on training data:", cv_auc.round(4), "| mean =", cv_auc.mean().round(4))

# no single feature explains the target on its own -> no target leakage
single_auc = {}
for col in num_cols:
    ok = X[col].notna()
    a = roc_auc_score(y[ok], X.loc[ok, col])
    single_auc[col] = max(a, 1 - a)
print("\nBest single-feature AUC:", max(single_auc, key=single_auc.get), round(max(single_auc.values()), 3))
```

**Interpretation:** cross-validation gives the same near-perfect AUC, and no single feature gives the answer alone (best single feature is far from 1.0). So there is **no data leakage**: this dataset is simply almost *linearly separable* (the risk label is a clean combination of several features such as income, debt, credit utilisation and late-payment days). This looks like a synthetic/practice dataset, so real-world scores would be lower.

### Task 12: Type-I and Type-II errors from the confusion matrix

```python
r = results["Logistic Regression (baseline)"]
print(f"True Negatives  (TN): {r['TN']}  -> low-risk correctly cleared")
print(f"True Positives  (TP): {r['TP']}  -> high-risk correctly caught")
print(f"Type-I  Error (FP)  : {r['FP (Type-I)']}  -> low-risk customers wrongly flagged")
print(f"Type-II Error (FN)  : {r['FN (Type-II)']}  -> HIGH-RISK customers MISSED (costly!)")
```

## Part D: Handling Imbalanced Data

### Task 13: Impact of class imbalance


```python
dummy = DummyClassifier(strategy="most_frequent").fit(X_train_p, y_train)
dummy_pred = dummy.predict(X_test_p)

print("Dummy model (always predicts Low Risk)")
print("  Accuracy :", round(accuracy_score(y_test, dummy_pred), 3))
print("  Recall   :", round(recall_score(y_test, dummy_pred), 3), " <- catches NO high-risk customer")
print()
print("Logistic Regression (baseline)")
print("  Accuracy :", round(results['Logistic Regression (baseline)']['Accuracy'], 3))
print("  Recall   :", round(results['Logistic Regression (baseline)']['Recall (High Risk)'], 3))
```

**Interpretation:** the useless dummy model already reaches a high accuracy, which proves accuracy is a poor metric on imbalanced data. We must look at **recall, F1 and AUC-ROC** for the minority class.



### Task 14: Apply Under-Sampling, Over-Sampling, SMOTE and ADASYN
> Resampling is applied **only on the training data**. The test data is never touched.

| Technique | Idea |
|---|---|
| Under-Sampling | randomly remove majority-class rows |
| Over-Sampling | randomly duplicate minority-class rows |
| SMOTE | create *synthetic* minority rows between neighbours |
| ADASYN | like SMOTE but creates more samples in harder-to-learn areas |

```python
samplers = {
    "Under-Sampling": RandomUnderSampler(random_state=42),
    "Over-Sampling":  RandomOverSampler(random_state=42),
    "SMOTE":          SMOTE(random_state=42),
    "ADASYN":         ADASYN(random_state=42),
}

balanced_data = {}
print(f"{'Original':<16} -> Low: {(y_train==0).sum():>5} | High: {(y_train==1).sum():>5}")
for name, sampler in samplers.items():
    X_res, y_res = sampler.fit_resample(X_train_p, y_train)
    balanced_data[name] = (X_res, y_res)
    print(f"{name:<16} -> Low: {(np.array(y_res)==0).sum():>5} | High: {(np.array(y_res)==1).sum():>5}")
```

```python
# Retrain Logistic Regression on each balanced training set
for name, (X_res, y_res) in balanced_data.items():
    model = LogisticRegression(max_iter=1000, random_state=42).fit(X_res, y_res)
    evaluate(f"LogReg + {name}", model)

balance_cols = ["Precision", "Recall (High Risk)", "F1-Score", "AUC-ROC", "FP (Type-I)", "FN (Type-II)"]
```

### Task 15: Compare performance before and after balancing

```python
before_after = pd.DataFrame(results).T.loc[
    ["Logistic Regression (baseline)"] + [f"LogReg + {n}" for n in samplers], balance_cols]
display(before_after.round(3))

# Bar chart: Recall, F1, AUC
plot_df = before_after[["Recall (High Risk)", "F1-Score", "AUC-ROC"]].copy()
plot_df.index = ["Before (baseline)"] + list(samplers)
plot_df.plot(kind="bar", figsize=(9, 4.5), colormap="viridis", width=0.8)
plt.title("Effect of Balancing Techniques (Logistic Regression)")
plt.ylabel("Score"); plt.ylim(0, 1); plt.xticks(rotation=20)
plt.legend(loc="lower right")
plt.show()
```

**Interpretation (full feature set):** the baseline already has recall ≈ 1.0, so balancing has nothing left to improve. It only adds a few false positives (Type-I errors), which lowers precision slightly. AUC-ROC is unchanged.

#### Stress test: imbalance effect on a harder problem (4 weaker features)
We keep only `annual_income_inr`, `debt_balance_inr`, `complaints_last_6m` and `age`, so the model can no longer separate the classes perfectly.

```python
hard_features = ["annual_income_inr", "debt_balance_inr", "complaints_last_6m", "age"]
idx = [i for i, f in enumerate(feature_names) if f.split("__")[-1] in hard_features]
Xtr_h, Xte_h = X_train_p[:, idx], X_test_p[:, idx]

stress = {}
variants = {"No balancing": None, **samplers}
for name, sampler in variants.items():
    Xr, yr = (Xtr_h, y_train) if sampler is None else sampler.fit_resample(Xtr_h, y_train)
    m = LogisticRegression(max_iter=1000, random_state=42).fit(Xr, yr)
    pred, proba = m.predict(Xte_h), m.predict_proba(Xte_h)[:, 1]
    stress[name] = {"Accuracy": accuracy_score(y_test, pred), "Precision": precision_score(y_test, pred),
                    "Recall (High Risk)": recall_score(y_test, pred), "F1-Score": f1_score(y_test, pred),
                    "AUC-ROC": roc_auc_score(y_test, proba)}
stress_df = pd.DataFrame(stress).T
display(stress_df.round(3))
stress_df.round(4).to_csv("stress_test_results.csv")

stress_df[["Accuracy", "Recall (High Risk)", "F1-Score", "AUC-ROC"]].plot(kind="bar", figsize=(9.5, 4.5), colormap="viridis", width=0.8)
plt.title("Stress Test (4 weak features): Before vs After Balancing")
plt.ylabel("Score"); plt.ylim(0, 1.05); plt.xticks(rotation=20); plt.legend(loc="lower right", fontsize=8)

plt.show()
```

**Interpretation (stress test):** this is the classic imbalance pattern.
- Without balancing: high accuracy but **low recall** (many Type-II errors, risky customers missed).
- With Under-Sampling, Over-Sampling, SMOTE and ADASYN: **recall rises a lot**, but precision and accuracy fall (more Type-I errors).
- AUC-ROC stays about the same, because balancing moves the decision threshold rather than making the model rank customers better.
- ADASYN usually gives the highest recall but the lowest precision; SMOTE / Over-Sampling give a gentler trade-off.

## Part E: Tree-Based Classification Models

### Task 16 & 17: Decision Tree and overfitting check

```python
dt = DecisionTreeClassifier(random_state=42)     
dt.fit(X_train_p, y_train)

def train_vs_test(name, model):
    tr, te = model.predict(X_train_p), model.predict(X_test_p)
    return {"Model": name,
            "Train Accuracy": accuracy_score(y_train, tr), "Test Accuracy": accuracy_score(y_test, te),
            "Train Recall": recall_score(y_train, tr),     "Test Recall": recall_score(y_test, te),
            "Train F1": f1_score(y_train, tr),             "Test F1": f1_score(y_test, te)}

evaluate("Decision Tree", dt)
overfit_table = pd.DataFrame([train_vs_test("Decision Tree", dt)]).set_index("Model")
display(overfit_table.round(3))
```

**Interpretation:** the unrestricted Decision Tree scores almost perfectly on training data but much worse on test data. It memorised the training set, which is **overfitting**.

### Task 18: Random Forest

```python
rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_train_p, y_train)
evaluate("Random Forest", rf)
print("Random Forest trained")
```

### Task 19: Decision Tree vs Random Forest

```python
tree_compare = pd.DataFrame([train_vs_test("Decision Tree", dt), train_vs_test("Random Forest", rf)]).set_index("Model")
tree_compare["Test AUC-ROC"] = [results["Decision Tree"]["AUC-ROC"], results["Random Forest"]["AUC-ROC"]]
tree_compare["Train-Test Accuracy Gap"] = tree_compare["Train Accuracy"] - tree_compare["Test Accuracy"]
display(tree_compare.round(3))

tree_compare[["Train Accuracy", "Test Accuracy"]].plot(kind="bar", figsize=(5.5, 4), color=["#6FA8DC", "#E69138"])
plt.title("Train vs Test Accuracy"); plt.ylim(0.5, 1.02); plt.xticks(rotation=0)
plt.show()
```

**Interpretation:** the Random Forest averages many trees built on random subsets, which reduces variance. Compared with the single Decision Tree it **generalises much better** (smaller train-test gap, higher test recall and AUC). Both reach 100% training accuracy, but only the forest keeps that performance on unseen data.

## Part F: Hyperparameter Tuning

We use **F1-score** as the tuning metric (it balances catching risky customers with not flagging everybody) and 5-fold stratified cross-validation (3-fold for the forest to keep it fast).

### Task 20: Randomized Search CV (Decision Tree & Random Forest)

```python

dt_params = {"max_depth": [3, 4, 5, 6, 8, 10, 15, None],
             "min_samples_split": [2, 5, 10, 20, 50],
             "min_samples_leaf": [1, 2, 5, 10, 20],
             "criterion": ["gini", "entropy"],
             "class_weight": [None, "balanced"]}

dt_search = RandomizedSearchCV(DecisionTreeClassifier(random_state=42), dt_params,
                               n_iter=25, scoring="f1", cv=StratifiedKFold(5, shuffle=True, random_state=42),
                               random_state=42, n_jobs=-1)
dt_search.fit(X_train_p, y_train)
print("Best Decision Tree params:", dt_search.best_params_)
print("Best CV F1:", round(dt_search.best_score_, 3))
dt_tuned = dt_search.best_estimator_
evaluate("Decision Tree (tuned)", dt_tuned)
```

```python

rf_params = {"n_estimators": [100, 200, 300],
             "max_depth": [5, 8, 12, 20, None],
             "min_samples_split": [2, 5, 10],
             "min_samples_leaf": [1, 2, 4, 8],
             "max_features": ["sqrt", "log2"],
             "class_weight": [None, "balanced"]}

rf_search = RandomizedSearchCV(RandomForestClassifier(random_state=42, n_jobs=-1), rf_params,
                               n_iter=15, scoring="f1", cv=StratifiedKFold(3, shuffle=True, random_state=42),
                               random_state=42, n_jobs=1)
rf_search.fit(X_train_p, y_train)
print("Best Random Forest params:", rf_search.best_params_)
print("Best CV F1:", round(rf_search.best_score_, 3))
rf_random = rf_search.best_estimator_
evaluate("Random Forest (random search)", rf_random)
```

### Task 21: Grid Search CV to fine-tune the best model

We build a small grid **around** the best values found by the random search and let Grid Search pick the final combination.

```python
best = rf_search.best_params_

def around(value, step, low=1):
    """Small neighbourhood around a numeric value."""
    return sorted({max(low, value - step), value, value + step})

grid = {"n_estimators": around(best["n_estimators"], 100, low=50),
        "max_depth": [None] if best["max_depth"] is None else around(best["max_depth"], 3),
        "min_samples_leaf": around(best["min_samples_leaf"], 1),
        "min_samples_split": [best["min_samples_split"]],
        "max_features": [best["max_features"]],
        "class_weight": [best["class_weight"]]}
print("Grid:", grid)

rf_grid = GridSearchCV(RandomForestClassifier(random_state=42, n_jobs=-1), grid,
                       scoring="f1", cv=StratifiedKFold(3, shuffle=True, random_state=42), n_jobs=1)
rf_grid.fit(X_train_p, y_train)
print("Best params after Grid Search:", rf_grid.best_params_)
print("Best CV F1:", round(rf_grid.best_score_, 3))
rf_tuned = rf_grid.best_estimator_
evaluate("Random Forest (tuned)", rf_tuned)
```

### Task 22: Tuned vs untuned performance

```python
tune_names = ["Decision Tree", "Decision Tree (tuned)", "Random Forest", "Random Forest (tuned)"]
tune_table = pd.DataFrame(results).T.loc[tune_names, ["Accuracy", "Precision", "Recall (High Risk)", "F1-Score", "AUC-ROC", "FN (Type-II)"]]
display(tune_table.round(3))

tune_table[["Recall (High Risk)", "F1-Score", "AUC-ROC"]].plot(kind="bar", figsize=(8, 4.5), colormap="viridis", width=0.8)
plt.title("Tuned vs Untuned Models"); plt.ylim(0, 1); plt.xticks(rotation=15)

plt.show()

print("Train vs Test check for the tuned models")
display(pd.DataFrame([train_vs_test("Decision Tree (tuned)", dt_tuned),
                      train_vs_test("Random Forest (tuned)", rf_tuned)]).set_index("Model").round(3))
```

**Interpretation:** tuning helps the single **Decision Tree a lot** (limiting depth and leaf size removes most of the overfitting, so recall and AUC jump). The **Random Forest was already near-perfect**, so tuning gives little or no visible gain; differences of 0 or 1 customer on the test set are just noise.

### Bonus: Tuned Random Forest trained on SMOTE data
Part D showed balancing raises recall. We test the same idea on the tuned forest so we can compare it fairly.

```python
X_sm, y_sm = balanced_data["SMOTE"]
rf_smote = RandomForestClassifier(**rf_grid.best_params_, random_state=42, n_jobs=-1).fit(X_sm, y_sm)
evaluate("Random Forest (tuned) + SMOTE", rf_smote)
display(pd.DataFrame(results).T.loc[["Random Forest (tuned)", "Random Forest (tuned) + SMOTE"],
        ["Precision", "Recall (High Risk)", "F1-Score", "AUC-ROC", "FP (Type-I)", "FN (Type-II)"]].round(3))
```

### Task 25: Select the best final model (business requirement: minimise false negatives)
Business rule used:
1. A model must still be **useful** → AUC-ROC within 0.03 of the best AUC (it must rank risky customers well).
2. Among those, choose the one with the **highest recall** (fewest missed high-risk customers).
3. Ties are broken by the better F1-score. If models are still tied, prefer the **simpler, more explainable** one.

```python
model_lookup = {"Random Forest (tuned)": rf_tuned, "Random Forest (tuned) + SMOTE": rf_smote,
                "Random Forest": rf, "Random Forest (random search)": rf_random,
                "Decision Tree": dt, "Decision Tree (tuned)": dt_tuned,
                "Logistic Regression (baseline)": log_reg}
for n, (Xr, yr) in balanced_data.items():
    model_lookup[f"LogReg + {n}"] = LogisticRegression(max_iter=1000, random_state=42).fit(Xr, yr)

final_model = model_lookup[best_name]
plot_confusion(final_model, f"Final model: {best_name}", "07_final_model_confusion_matrix")
print(classification_report(y_test, final_model.predict(X_test_p), target_names=["Low Risk", "High Risk"]))
```

#### Business interpretation of False Positives and False Negatives
- **False Positive (Type-I):** a safe customer is flagged. Cost = a manual review or a short call to the customer. It is cheap, but too many hurt customer experience and waste analyst time.
- **False Negative (Type-II):** a risky customer is missed. Cost = unpaid loans, chargebacks or fraud losses. This is far more expensive, so the bank is willing to accept some extra false positives to catch more risky customers.

#### Impact of imbalance-handling techniques
- A model that always says "Low Risk" gets ~88% accuracy but 0% recall, so accuracy is not a valid metric here.
- On the **full feature set** the data is almost perfectly separable, so balancing is not needed; it only adds a few false positives.
- On the **harder 4-feature stress test**, balancing (Under-Sampling, Over-Sampling, SMOTE, ADASYN) raises recall sharply (fewer missed risky customers) at the cost of precision. AUC-ROC does not change.
- Class weights (`class_weight="balanced"`) inside the tree models work in a similar way.

#### Conclusions & Recommendations
1. **Selected model:** the one printed above by the business rule (minimum missed high-risk customers). Logistic Regression and the tuned Random Forest are practically tied at the top; Logistic Regression is simpler and easier to explain, while the Random Forest gives feature importances. A single Decision Tree overfits and is the weakest.
2. Judge models with **Recall, F1 and AUC-ROC**, never accuracy alone.
3. The test set has only ~110 high-risk customers, so a difference of 1 or 2 customers between models is **not statistically meaningful**.
4. Near-perfect scores are unusual for real banking data. Before production, **validate on fresh, real data**, and if scores are lower, use the balancing techniques (SMOTE / class weights) shown in the stress test.
5. Send flagged customers to a **review queue** instead of automatic rejection, because some flags will be false alarms.
6. **Monitor** the model every month and retrain when customer behaviour changes. Next steps: tune the probability threshold and try Gradient Boosting / XGBoost.
